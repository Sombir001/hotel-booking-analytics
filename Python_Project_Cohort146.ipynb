### 📝 Initial Data-Quality Assessment

- **Missing Observations**
  - `children`: Some records have missing values; absence generally represents no recorded children.
  - `country`: Missing entries are marked as **Unknown** (no invented country values).
  - `agent` and `company`: Missing IDs treated as **0** (no recorded agent/company), since they are identifiers and not continuous measurements.
  - `company`: Contains particularly high missingness.

- **Duplicates**
  - Duplicate observations were identified and removed.

---

✅ **Action Taken**: Missing values and duplicates are handled before conducting the main analysis.


```python
hotel = df.copy()
```

### Handling Missing Data

- **children = 0**  
  Reasonable because absence generally represents no recorded children.

- **country = Unknown**  
  Do not invent a country; mark as "Unknown".

- **agent/company = 0**  
  Treat missing IDs as no recorded agent/company rather than replacing them with a mean.


```python
hotel["children"] = hotel["children"].fillna(0)

hotel["country"] = hotel["country"].fillna("Unknown")

hotel["agent"] = hotel["agent"].fillna(0)

hotel["company"] = hotel["company"].fillna(0)
```

```python
print("Missing values remaining:")
display(hotel.isnull().sum().sort_values(ascending=False).head(10))
```

```python
hotel["children"] = hotel["children"].astype(int)

hotel["agent"] = hotel["agent"].astype(int)

hotel["company"] = hotel["company"].astype(int)

hotel["reservation_status_date"] = pd.to_datetime(
    hotel["reservation_status_date"]
)
```

```python
before = len(hotel)

hotel = hotel.drop_duplicates().copy()

after = len(hotel)

print(f"Rows before removing duplicates : {before:,}")
print(f"Rows after removing duplicates  : {after:,}")
print(f"Duplicates removed              : {before-after:,}")
```

Total length of stay

```python
hotel["total_stay"] = (
    hotel["stays_in_weekend_nights"] +
    hotel["stays_in_week_nights"]
)
```

Total guests

```python
hotel["total_guests"] = (
    hotel["adults"] +
    hotel["children"] +
    hotel["babies"]
)
```

Has children

```python
hotel["has_children"] = np.where(
    (hotel["children"] > 0) | (hotel["babies"] > 0),
    "Yes",
    "No"
)
```

Special-request indicator

```python
hotel["has_special_request"] = np.where(
    hotel["total_of_special_requests"] > 0,
    "Yes",
    "No"
)
```

Create actual arrival date

```python
hotel["arrival_date"] = pd.to_datetime(
    hotel["arrival_date_year"].astype(str) + "-" +
    hotel["arrival_date_month"] + "-" +
    hotel["arrival_date_day_of_month"].astype(str),
    errors="coerce"
)
```

Day of week

```python
hotel["arrival_day"] = hotel["arrival_date"].dt.day_name()
```

Q1: What is the average lead time for bookings that are cancelled versus those that are not cancelled?

```python
q1 = (
    hotel.groupby("is_canceled")["lead_time"]
    .mean()
    .rename(index={0: "Not Cancelled", 1: "Cancelled"})
    .round(2)
)

display(q1.to_frame("Average Lead Time"))
```

Subjective Questions:

Q1:	How does the lead time affect the likelihood of booking cancellations?

```python
hotel["lead_time_group"] = pd.cut(
    hotel["lead_time"],
    bins=[-1, 7, 30, 90, 180, 365, np.inf],
    labels=[
        "0-7 days",
        "8-30 days",
        "31-90 days",
        "91-180 days",
        "181-365 days",
        "366+ days"
    ]
)

lead_cancel = (
    hotel.groupby(
        "lead_time_group",
        observed=True
    )["is_canceled"]
    .mean()
    .mul(100)
    .reset_index(name="Cancellation Rate")
)

display(lead_cancel)
```

```python
plt.figure(figsize=(11, 6))

sns.barplot(
    data=lead_cancel,
    x="lead_time_group",
    y="Cancellation Rate"
)

plt.title("Cancellation Rate by Lead-Time Group")
plt.xlabel("Lead-Time Group")
plt.ylabel("Cancellation Rate (%)")

plt.xticks(rotation=30)
plt.show()
```

Cancellation probability rises considerably as lead time increases

2.	What impact do special requests have on booking cancellations?

```python
request_cancel = (
    hotel.groupby("has_special_request")["is_canceled"]
    .mean()
    .mul(100)
    .round(2)
)

display(
    request_cancel.to_frame("Cancellation Rate (%)")
)
```

Guests making special requests have a noticeably lower cancellation rate. Special requests may therefore act as a useful signal of stronger booking commitment or engagement.

3.	What trends can be observed in the booking behavior of international guests compared to domestic guests?

```python
hotel["guest_origin"] = np.where(
    hotel["country"].eq("PRT"),
    "Domestic",
    np.where(
        hotel["country"].eq("Unknown"),
        "Unknown",
        "International"
    )
)

origin_analysis = hotel.groupby("guest_origin").agg(
    Bookings=("hotel", "size"),
    Average_ADR=("adr", "mean"),
    Average_Lead_Time=("lead_time", "mean"),
    Average_Stay=("total_stay", "mean"),
    Cancellation_Rate=("is_canceled", "mean")
)

origin_analysis["Cancellation_Rate"] *= 100

display(origin_analysis.round(2))
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=hotel[hotel["guest_origin"] != "Unknown"],
    x="guest_origin",
    y="lead_time",
    estimator="mean"
)

plt.title("Average Lead Time: Domestic vs International Guests")
plt.xlabel("Guest Origin")
plt.ylabel("Average Lead Time")

plt.show()
```

## 📊 Key Insights and Recommendations

- **Lead Time & Cancellation Risk**  
  - Bookings made within **7 days of arrival** show substantially lower cancellation rates.  
  - Long-lead bookings (months in advance) have higher cancellation risk.  
  - 🔑 *Recommendation*: Hotels should consider targeted reconfirmation messages or flexible deposit strategies for long-lead bookings.

- **Special Requests & Booking Commitment**  
  - Guests with special requests tend to cancel less.  
  - Personalization indicates stronger booking commitment.  
  - 🔑 *Recommendation*: Encourage pre-arrival engagement and personalization; use request activity as an indicator of booking stability.

- **Seasonality (August Peak)**  
  - August records the largest booking volume, showing a clear seasonal peak.  
  - 🔑 *Recommendation*: Adjust staffing, inventory, room availability, and pricing to anticipate higher demand in August.

- **Repeat vs. New Guests**  
  - Repeat guests have shorter lead times compared to new guests.  
  - 🔑 *Recommendation*: Loyalty-oriented communication and targeted last-minute offers may be especially effective for returning customers.

- **Market Segments & Distribution Channels**  
  - Cancellation rates vary across segments and channels.  
  - 🔑 *Recommendation*: Evaluate channels not only on booking volume, but also on cancellation behavior, ADR, and demand quality.
